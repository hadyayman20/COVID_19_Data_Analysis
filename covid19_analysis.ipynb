## 3. Data Cleaning and Preprocessing

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
global_locations = ["World", "International"]
df_countries = df[~df["location"].isin(global_locations)].copy()
df_world = df[df["location"] == "World"].copy()
```

```python
missing = df_countries.isna().mean()
df_clean = df_countries.drop(columns=missing[missing > 0.60].index)
df_clean = df_clean.sort_values(["location", "date"])
```

```python
cumulative_cols = ["total_cases", "total_deaths",
                   "total_cases_per_million", "total_deaths_per_million"]

for col in cumulative_cols:
    df_clean[col] = df_clean.groupby("location")[col].ffill()
```

```python
daily_cols = ["new_cases", "new_deaths",
              "new_cases_smoothed", "new_deaths_smoothed",
              "new_cases_per_million", "new_deaths_per_million"]

for col in daily_cols:
    df_clean[col] = df_clean[col].fillna(0)
```

```python
static_cols = ["population", "population_density", "median_age", "aged_65_older", "aged_70_older",
               "gdp_per_capita", "cardiovasc_death_rate", "diabetes_prevalence",
               "hospital_beds_per_thousand", "life_expectancy", "human_development_index"]

for col in static_cols:
    median_values = df_clean.groupby("location")[col].transform("median")
    df_clean[col] = df_clean[col].fillna(median_values)
```

```python
df_clean["month"] = df_clean["date"].dt.to_period("M").astype(str)
df_clean["week"] = df_clean["date"].dt.strftime("%U")
df_clean["death_rate"] = (
    df_clean["total_deaths"] / df_clean["total_cases"] * 100
)
```

```python
print(df_clean.shape)
df_clean.head()
```

```python
day_wise = df_clean.groupby("date")[[
    "total_cases",
    "total_deaths",
    "new_cases",
    "new_deaths"
]].sum()

day_wise["No. of countries"] = df_clean.groupby("date")["location"].nunique()

day_wise["Deaths / 100 Cases"] = (
    day_wise["total_deaths"] / day_wise["total_cases"] * 100
).round(2)

day_wise = day_wise.reset_index()

day_wise.tail()
```

```python
latest_date = df_clean["date"].max()

country_wise = df_clean[df_clean["date"] == latest_date].copy()

country_wise["death_rate"] = (
    country_wise["total_deaths"] /
    country_wise["total_cases"] * 100
).round(2)

print("Snapshot date:", latest_date.date())

country_wise.head()
```

```python
total_cases = country_wise["total_cases"].sum()
total_deaths = country_wise["total_deaths"].sum()

print("Total Cases:", int(total_cases))
print("Total Deaths:", int(total_deaths))
```

```python
top10_cases = country_wise.sort_values("total_cases", ascending=False).head(10)
top10_cases[["location", "total_cases", "total_deaths"]]
```

## 4. Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=top10_cases, x="total_cases", y="location", color="royalblue")
plt.title("Top 10 Countries by Confirmed Cases")
plt.xlabel("Total Cases")
plt.ylabel("Countries")
plt.show()
```

**Insight:** The United States, India, and Brazil are far ahead of every other country in total confirmed cases. Countries with very large populations (US, India, Brazil, Russia) dominate this ranking, which is expected.

```python
plt.figure(figsize=(12, 5))
plt.plot(day_wise["date"], day_wise["new_cases"], color="steelblue")
plt.title("New Cases Over Time")
plt.xlabel("Date")
plt.ylabel("New Cases")
plt.show()
```

**Insight:** New cases were still climbing through the whole period covered by the dataset. the highest single-day total (~314,000 new cases) was recorded on the very last date in the data. This suggests the pandemic
was still in an active growth phase worldwide at this time.

```python
plt.figure(figsize=(12, 5))
plt.plot(day_wise["date"], day_wise["new_deaths"], color="crimson")
plt.title("New Deaths Over Time")
plt.xlabel("Date")
plt.ylabel("New Deaths")
plt.show()
```

**Insight:** Unlike new cases, new deaths peaked much earlier (mid-April 2020, at around 10,500 deaths in a single day) and then dropped even as new cases kept rising through the summer. This gap between rising cases and falling deaths is consistent with expanded testing and improved treatment over time.

```python
top5_countries = top10_cases["location"].head(5).tolist()

plt.figure(figsize=(12, 6))
for country in top5_countries:
    data = df_clean[df_clean["location"] == country]
    plt.plot(data["date"], data["total_cases"], label=country)

plt.title("Total Top 5 Countries Cases Over Time ")
plt.xlabel("Date")
plt.ylabel("Total Cases")
plt.legend()
plt.show()
```

**Insight:** The United States shows a much earlier climb than the other top 5 countries, India's curve starts later but grows the fastest of all by the end of the period despite starting its major outbreak months later.

```python
cols = ["total_cases", "total_deaths", "population", "gdp_per_capita",
        "median_age", "life_expectancy", "human_development_index"]

corr = country_wise[cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Variables")
plt.show()
```

**Insight:** `total_deaths` is very strongly correlated with `total_cases` (as expected since more infections means more deaths). `population` also correlates with `total_cases`.

```python
week_wise = df_clean.groupby("week")[["new_cases", "new_deaths"]].sum().reset_index()
week_wise.head()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=week_wise, x="week", y="new_cases", color="teal")
plt.title("New Cases per Week")
plt.xlabel("Week Number")
plt.ylabel("New Cases")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=week_wise, x="week", y="new_deaths", color="crimson")
plt.title("New Deaths per Week")
plt.xlabel("Week Number")
plt.ylabel("New Deaths")
plt.xticks(rotation=90)
plt.show()
```

**Insight:** New cases rise steadily on a week-by-week basis. New deaths in contrast, spike hard in the early weeks and then drops out at a lower, steady level for the rest of the period.

```python
month_wise = df_clean.groupby("month")[["new_cases", "new_deaths"]].sum().reset_index()
month_wise
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=month_wise, x="month", y="new_cases", color="teal")
plt.title("New Cases per Month")
plt.xlabel("Month")
plt.ylabel("New Cases")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=month_wise, x="month", y="new_deaths", color="crimson")
plt.title("New Deaths per Month")
plt.xlabel("Month")
plt.ylabel("New Deaths")
plt.show()
```

**Insight:** Monthly new cases show a clear acceleration climbing every month afterward and peaking in August 2020 (~8 million new cases) before dropped slightly in September. Monthly new deaths in cosntrat peaked in
April 2020 (Around 189,000 deaths that month).

## 5. Conclusions and Insights

Based on the analysis above:

- **Cases are concentrated:** the United States, India, and Brazil alone account for the majority of global confirmed cases, and all of the top-10 countries by case count are large, populous nations. raw case totals are heavily driven by population size, not just how well or badly a country handled the outbreak.
- **Deaths peaked earlier than cases:** global new deaths peaked in mid-April 2020 and declined afterward, while new cases kept climbing every month through August 2020. This gap likely reflects wider testing and better treatment.
- **Fatality rate varies a lot from a country to another:** even among the top-10 countries by cases, the death rate ranges from under 2% to over 10% total case count alone doesn't predict how deadly the outbreak was in a given country.
- **Country wealth and age aren't strong predictors:** indicators like GDP per capita, median age, and life expectancy showed only weak correlation with total cases, suggesting factors like population size, government response, and healthcare capacity mattered more than general development level.
